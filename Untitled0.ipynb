```python
import folium
from folium.plugins import HeatMap
import pandas as pd

# 1. Örnek Veri Setimiz (Lokasyonlar ve Fiyat/Ağırlık Değerleri)
veri = {
    'enlem': [40.1020, 40.1150, 40.1300, 40.1480, 40.0600, 40.1100, 40.1050, 40.1120],
    'boylam': [26.3920, 26.4080, 26.4020, 26.4050, 26.3500, 26.4000, 26.3950, 26.4060],
    'fiyat_tl': [12000, 18000, 14000, 25000, 32000, 22000, 16000, 21000]
}

df = pd.DataFrame(veri)

# 2. Haritayı Başlatıyoruz
harita_v4 = folium.Map(
    location=[40.1100, 26.3900],
    zoom_start=12,
    tiles='https://server.arcgisonline.com/ArcGIS/rest/services/World_Street_Map/MapServer/tile/{z}/{y}/{x}',
    attr='Esri'
)

# 3. Verileri [Enlem, Boylam, Ağırlık/Fiyat] formatına çeviriyoruz
isi_verisi = [[row['enlem'], row['boylam'], row['fiyat_tl']] for index, row in df.iterrows()]

# 4. HeatMap (Sıcaklık Katmanı) Ekliyoruz
HeatMap(isi_verisi, radius=30, blur=20, min_opacity=0.4).add_to(harita_v4)

# Haritayı Göster
harita_v4
```
